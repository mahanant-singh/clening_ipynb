# Pandas String Methods for Machine Learning

Machine Learning mein Pandas String Methods (`.str`) ka use **text cleaning, data preprocessing aur feature engineering** mein hota hai.

Examples mein `Company`, `Product`, `Specs` jaise text columns use kiye gaye hain.

```python
import pandas as pd

df = pd.DataFrame({
    "Company": [" DELL ", "HP", "LENOVO", "Dell"],
    "Product": ["Dell XPS", "HP Laptop Pro", "Lenovo ThinkPad", "Dell Inspiron"],
    "Specs": ["8GB RAM", "16GB RAM", "8GB 16GB options", "32GB RAM"],
    "Code": ["123", "A45", "456", "789"],
    "Name": ["hello", "WORLD", "mahanant", "Data Science"],
    "Text": ["banana", "apple", "machine learning", "pandas"]
})

df
```

## 1. `.str.lower()` — Text ko lowercase karna

```python
df["Company"].str.lower()
```

## 2. `.str.upper()` — Text ko uppercase karna

```python
df["Company"].str.upper()
```

## 3. `.str.title()` — Har word ka first letter capital karna

```python
df["Name"].str.title()
```

## 4. `.str.capitalize()` — Sirf first character capital karna

```python
df["Name"].str.capitalize()
```

## 5. `.str.strip()` — Starting aur ending spaces remove karna

```python
df["Company"].str.strip()
```

## 6. `.str.lstrip()` — Left side ke spaces remove karna

```python
df["Company"].str.lstrip()
```

## 7. `.str.rstrip()` — Right side ke spaces remove karna

```python
df["Company"].str.rstrip()
```

## 8. `.str.replace()` — Text ya pattern replace karna

```python
df["Product"].str.replace("HP ", "", regex=False)
```

## 9. `.str.contains()` — Specific word ya pattern check karna

`na=False` missing values ko `False` treat karta hai. `case=False` case-insensitive search karta hai.

```python
df["Product"].str.contains("dell", case=False, na=False)
```

### Example: Touchscreen feature banana

```python
screen = pd.Series(["1920x1080 Touchscreen", "1366x768", None])
touchscreen = screen.str.contains("touchscreen", case=False, na=False).astype(int)
touchscreen
```

## 10. `.str.startswith()` — Prefix se start hona check karna

```python
df["Product"].str.startswith("Dell", na=False)
```

## 11. `.str.endswith()` — Suffix par end hona check karna

```python
df["Product"].str.endswith("Pro", na=False)
```

## 12. `.str.split()` — Text ko parts mein divide karna

```python
df["Product"].str.split()
```

## 13. `.str.len()` — String ke characters count karna

```python
df["Product"].str.len()
```

## 14. `.str.get()` — Position se element lena

```python
df["Product"].str.split().str.get(0)
```

## 15. `.str.extract()` — Regex se matching group extract karna

```python
df["Specs"].str.extract(r"(\\d+)")
```

## 16. `.str.findall()` — Regex se sab matching values ki list nikalna

```python
df["Specs"].str.findall(r"\\d+GB")
```

## 17. `.str.isdigit()` — Check karna ki sab characters digits hain

```python
df["Code"].str.isdigit()
```

## 18. `.str.isalpha()` — Check karna ki sab characters letters hain

```python
df["Name"].str.isalpha()
```

## 19. `.str.isnumeric()` — Check karna ki sab characters numeric hain

```python
df["Code"].str.isnumeric()
```

## 20. `.str.count()` — Pattern kitni baar aaya hai

```python
df["Text"].str.count("a")
```

## 21. `.str.slice()` — String ka selected portion nikalna

```python
df["Product"].str.slice(0, 3)
```

## Quick revision

| Method | Common ML use |
|---|---|
| `.str.strip()` | Extra spaces clean karna |
| `.str.lower()` | Text standardize karna |
| `.str.replace()` | Unwanted text/characters remove karna |
| `.str.contains()` | Boolean feature banana |
| `.str.split()` / `.str.get()` | Text se parts/features nikalna |
| `.str.extract()` | Numbers ya patterns extract karna |
| `.str.findall()` | Multiple matching values nikalna |
| `.str.len()` | Text length ko feature banana |

**Note:** Missing values wale columns par `na=False` ya suitable missing-value handling use karein. Har string method har column ke liye zaroori nahi hota; pehle data inspect karein.
